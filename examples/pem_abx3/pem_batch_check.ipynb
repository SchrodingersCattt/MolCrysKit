# ABX3 PEM × MolCrysKit 批量检查
- 来源：Guo et al., EMF 2026 (DOI:10.1016/j.enmf.2026.03.001)
- 目标：批量读取当前文件夹下所有 CIF，识别分子/空间群/无序
- 运行日期：2026-09-21

```python
import os
from glob import glob
import molcrys_kit as mck
import pandas as pd

print("molcrys_kit 版本:", mck.__version__)
print("pandas 版本:", pd.__version__)
```

```python
# 当前文件夹下所有 .cif 文件
cif_files = sorted(glob("*.cif"))

print(f"找到 {len(cif_files)} 个 CIF 文件:\n")
for f in cif_files:
    print(" -", f)
```

```python
rows = []

for path in cif_files:
    name = os.path.basename(path)
    print(f"\n{'='*50}")
    print(f"正在处理: {name}")
    try:
        # 先试 read_mol_crystal
        try:
            crystal = mck.read_mol_crystal(path)
        except Exception:
            # 不行就换 from_cif
            crystal = mck.MolecularCrystal.from_cif(path)
        
        formulas = []
        for i, mol in enumerate(crystal.molecules):
            formula = mol.get_chemical_formula()
            formulas.append(formula)
            print(f"  分子 {i}: {formula}")
        
        lat = crystal.get_lattice_parameters()
        print(f"  晶胞参数: {lat}")
        print(f"  共 {len(crystal.molecules)} 个分子/离子单元")
        
        rows.append({
            "file": name,
            "status": "OK",
            "n_molecules": len(crystal.molecules),
            "formulas": " | ".join(formulas),
            "lattice": str(lat),
            "error": ""
        })
    except Exception as e:
        print(f"  ❌ 失败: {e}")
        rows.append({
            "file": name,
            "status": "FAIL",
            "n_molecules": None,
            "formulas": "",
            "lattice": "",
            "error": str(e)
        })

print(f"\n{'='*50}")
ok_count = sum(1 for r in rows if r['status'] == 'OK')
fail_count = sum(1 for r in rows if r['status'] == 'FAIL')
print(f"完成！成功 {ok_count} 个，失败 {fail_count} 个")
```

```python
df = pd.DataFrame(rows, columns=[
    "file", "status", "n_molecules", "formulas", "lattice", "error"
])
df
```

```python
# 保存成 markdown 表
with open("batch_run.md", "w", encoding="utf-8") as f:
    f.write("# MolCrysKit 批量运行结果\n\n")
    f.write(df.to_markdown(index=False))

# 也保存成 csv（Excel 能直接打开）
df.to_csv("batch_run.csv", index=False, encoding="utf-8-sig")

print("已保存:")
print(" - batch_run.md")
print(" - batch_run.csv")
```
